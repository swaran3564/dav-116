```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

d = load_iris()
df = pd.DataFrame(d.data, columns=d.feature_names)
df["species"] = [d.target_names[i] for i in d.target]

print("Aggregation:")
print(df.groupby("species")["petal length (cm)"].mean().round(2))

print("\nFiltered Data:")
print(df[df["petal length (cm)"] > 5].head())

print("\nCorrelation:")
print(df.iloc[:, :4].corr().round(2))

print("\nCovariance:")
print(df.iloc[:, :4].cov().round(2))

plt.scatter(df["sepal length (cm)"],
            df["petal length (cm)"], c=d.target)
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.title("Iris Scatter Plot")
plt.show()
```
